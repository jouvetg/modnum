# Se familiariser avec les figures interactives

Les figures interactives permettent de visualiser l'évolution d'un modèle numérique **en temps réel**. Le principe : on crée la figure **avant** la boucle, puis on la redessine à chaque pas de temps.

Trois commandes sont essentielles **dans** la boucle :

1. **`clear_output(wait=True)`** : nettoie la sortie précédente, pour éviter la superposition ;
2. **`ax.cla()`** (ou son synonyme `ax.clear()`) : efface les éléments visuels précédents du tracé ;
3. **`display(fig)`** : affiche la figure mise à jour.

## 1. Utiliser `ax` plutôt que `plt`

Quand on définit `fig, ax = plt.subplots()`, `fig` est la fenêtre graphique globale et `ax` la zone où l'on trace. Les commandes passent alors par `ax` :

| Commande avec `plt` | Équivalent avec `ax` |
|---|---|
| `plt.plot(x, y)` | `ax.plot(x, y)` |
| `plt.xlabel('label')` | `ax.set_xlabel('label')` |
| `plt.title('titre')` | `ax.set_title('titre')` |
| `plt.ylim([a, b])` | `ax.set_ylim([a, b])` |

## 2. Fréquence d'affichage

L'affichage coûte cher. Si la fréquence est **trop faible** on ne voit pas l'évolution, **trop élevée** le code est ralenti inutilement. On affiche donc périodiquement, avec l'opérateur modulo : `if it % freq_affich == 0`.

## 3. Structure complète

```python
fig, ax = plt.subplots()               # AVANT la boucle

for it in range(nt):
    # ... calculs ...

    if it % freq_affich == 0:          # affichage periodique
        clear_output(wait=True)        # nettoie la sortie
        ax.cla()                       # efface le trace precedent
        ax.plot(x, y)                  # trace les nouvelles donnees
        ax.set_xlabel('...')
        display(fig)                   # affiche la figure
        plt.pause(0.1)                 # pause optionnelle
```

## 4. Exemple : une particule sur le cercle trigonométrique

````python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display

nt          = 300      # nombre total d'iterations
freq_affich = 10       # frequence d'affichage

fig, ax = plt.subplots(figsize=(4.5,4.5))     # figure creee AVANT la boucle
theta = np.linspace(0, 2*np.pi, 100)

for it in range(nt):

    x = np.cos(it*0.02)          # le "modele" : la particule avance
    y = np.sin(it*0.02)

    if it % freq_affich == 0:
        clear_output(wait=True)
        ax.cla()
        ax.plot(np.cos(theta), np.sin(theta), 'k--', alpha=0.3)
        ax.scatter(x, y, c='red', s=100)
        ax.set_xlim(-1.2, 1.2) ; ax.set_ylim(-1.2, 1.2)   # limites FIXES
        ax.set_aspect('equal')
        ax.set_title(f'Itération {it}/{nt}')
        display(fig) ; plt.pause(0.02)

plt.close()
````

## 5. Plusieurs sous-figures

On peut découper la fenêtre avec `fig, (ax1, ax2) = plt.subplots(2, 1)`. Il faut alors nettoyer **chaque** axe (`ax1.cla()`, `ax2.cla()`) et les remplir séparément.

**⚠️ Important** : en figure interactive, il est **essentiel de fixer les limites des axes** avec `ax.set_xlim()` et `ax.set_ylim()`. Sans cela, les axes se réajustent à chaque itération, ce qui crée un effet de « zoom » gênant et empêche de suivre l'évolution des données.

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6,4))
nt, freq_affich = 200, 10

for it in range(nt):

    t = np.linspace(0, it*0.005, it+1)

    if it % freq_affich == 0:
        clear_output(wait=True)
        ax1.cla() ; ax2.cla()
        ax1.plot(t, np.cos(t), 'b') ; ax1.set_ylabel('x(t)')
        ax2.plot(t, np.sin(t), 'r') ; ax2.set_ylabel('y(t)') ; ax2.set_xlabel('temps')
        for a in [ax1, ax2]:
            a.set_xlim([0, 1.0]) ; a.set_ylim([-1.2, 1.2]) ; a.grid(True)
        plt.tight_layout() ; display(fig) ; plt.pause(0.02)

plt.close()
````
