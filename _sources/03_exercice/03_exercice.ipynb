# Trajectoire d’une bombe volcanique

Une bombe volcanique est éjectée lors d'une éruption depuis une altitude de 480 m, à la vitesse de 300 km/h et avec un angle de 60 degrés par rapport à l'horizontale. Le volcan se dresse sur une **île** : l'altitude $y = 0$ correspond donc au **niveau de la mer**, où la bombe finit par retomber. Le but de l'exercice est de simuler sa trajectoire en discrétisant et résolvant les équations de la dynamique en 2D.

| **Paramètre**                      | **Valeur** | **Unité** |
|------------------------------------|------------|-----------|
| Position initiale, $(x_0,y_0)$     | (0, 480)   | m         |
| Vitesse d'éjection, $V$            | 300        | km/h      |
| Angle d'éjection, $\alpha$         | 60         | degrés    |
| Accélération de la pesanteur, $g$  | 9.81       | m/s²      |
| Durée totale, $t_\mathrm{tot}$     | 30         | s         |


## Solution "analytique"

En fait, ce problème particulier possède une **solution exacte** (ou **analytique**), que l'on peut écrire directement :

$$x(t) = x_0 + V\cos(\alpha)\, t, \qquad y(t) = y_0 + V\sin(\alpha)\, t - \frac{1}{2} g t^2.$$

En théorie, nous n'avons donc pas besion de calculer une solution numérique. Nous le ferons quand même en guise d'entrainement, et utiliserons la solution exacte/analytique pour vérifier.

La cellule ci-dessous implémente la solution exacte et la trace. **Exécutez-la** : c'est exactement la trajectoire que votre modèle numérique devra reproduire. Remarquez qu'aucune boucle n'est nécessaire — les deux formules sont évaluées sur 500 instants d'un coup.

````python
import numpy as np
import matplotlib.pyplot as plt

# parametres physiques
Xb_ini = 0            # position initiale en x, m
Yb_ini = 480          # position initiale en y, m
V_kmh  = 300          # vitesse d'ejection, km/h
V      = V_kmh / 3.6  # CONVERSION en m/s, pour etre coherent avec g et les positions
alpha  = 60           # angle d'ejection, degres
g      = 9.81         # pesanteur, m/s2
duree  = 30           # duree totale, s

Vx     = np.cos(np.radians(alpha)) * V   # composante horizontale de la vitesse, m/s
Vy_ini = np.sin(np.radians(alpha)) * V   # composante verticale initiale, m/s

t_sol = ( Vy_ini + np.sqrt(Vy_ini**2 + 2*g*Yb_ini) ) / g   # instant ou la bombe touche le sol

t       = np.linspace(0, t_sol, 500)
x_exact = Xb_ini + Vx * t
y_exact = Yb_ini + Vy_ini * t - 0.5 * g * t**2

plt.figure(figsize=(8,4))

# le decor : l'ile volcanique et la mer
plt.fill_between([-700, 900], -120, 0, color='lightblue')          # la mer
plt.fill([-600, 0, 600], [0, Yb_ini, 0], color='saddlebrown')      # le volcan
plt.axhline(0, color='navy', linewidth=1)
plt.text(0, Yb_ini+30, 'volcan', ha='center', color='saddlebrown')
plt.text(750, -75, 'niveau de la mer', ha='right', color='navy')

plt.plot(x_exact, y_exact, 'b')
plt.xlabel('distance horizontale, m')
plt.ylabel('altitude, m')
plt.title('Trajectoire attendue (solution analytique)')
plt.xlim([-700, 900])
plt.ylim([-120, 850])
plt.grid(True)
plt.show()

print("la bombe touche le sol apres", round(t_sol, 1), "s, a", round(x_exact[-1]), "m")
````

## 3. Et si l'on ajoute le vent ? *(facultatif)*

Jusqu'ici, la bombe ne subit que la pesanteur — et c'est bien pour cela qu'une solution exacte existe. Ajoutons maintenant un **vent horizontal**, qui emporte la bombe pendant son vol. Le vent n'est pas uniforme — **il est plus fort en altitude qu'au sol**. Prenons le modèle le plus simple, une croissance linéaire avec l'altitude :

$$V_\mathrm{vent}(y) = V_\mathrm{ref}\, \frac{y}{y_\mathrm{ref}}, \qquad V_\mathrm{ref} = 50~\mathrm{km/h} \ \text{ à } \ y_\mathrm{ref} = 480~\mathrm{m}.$$

**Ce qui change par rapport à la question 1 :** une seule des quatre équations. La bombe se déplace horizontalement à sa propre vitesse **plus celle du vent** :

$$\frac{dx}{dt} = V_x + V_\mathrm{vent}(y).$$

Les trois autres sont inchangées — une seule ligne de code à modifier.

Et pourtant, **il n'y a plus de solution analytique**. Cependant, le modèle numérique du code précédent s'en accommode sans difficulté.
